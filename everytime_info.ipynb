```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# ---------------------------------------------------------
# 1. 환경 설정 (이미 driver가 선언되어 있다면 이 부분은 제외해도 됩니다)
# driver = webdriver.Chrome() 
# ---------------------------------------------------------

def get_all_course_names_final(driver):
    """현재 화면에 로드된 모든 강의명을 추출하는 함수"""
    try:
        print("데이터 로딩 대기 중...")
        # '조회' 글자가 포함된 a 태그가 나타날 때까지 최대 10초 대기
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.XPATH, "//a[contains(text(), '조회')]"))
        )
        
        # '조회' 버튼이 있는 행(tr)의 4번째 칸(td)인 '강좌명'을 모두 찾음
        # 에타 구조상 td[4]가 강좌명입니다.
        course_elements = driver.find_elements(By.XPATH, "//tr[td/a[contains(text(), '조회')]]/td[4]")
        
        names = [el.text.strip() for el in course_elements if el.text.strip()]
        return names
    except Exception as e:
        print(f"오류 발생: {e}")
        return []

# ==========================================
# 메인 실행 로직
# ==========================================

# 1. 수집된 강의명을 저장할 집합 (중복 방지)
final_course_set = set()

print("!!! 필독: 브라우저에서 '산업경영공학과' 등 원하는 학과를 클릭한 상태여야 합니다.")

while True:
    choice = input("\n현재 페이지의 강의를 수집할까요? (y: 수집, n: 종료 후 크롤링 시작): ").lower()
    
    if choice == 'y':
        current_batch = get_all_course_names_final(driver)
        
        if not current_batch:
            print("❌ 강의를 찾지 못했습니다. 페이지가 다 로딩되었는지 확인해주세요.")
        else:
            final_course_set.update(current_batch)
            print(f"✅ 현재 페이지에서 {len(current_batch)}개 수집 완료!")
            print(f"현재까지 총 누적 강의 수: {len(final_course_set)}")
            print(f"수집된 샘플: {list(final_course_set)[:3]}...")
            
    elif choice == 'n':
        break

# 2. 수집 완료 후 기존 크롤링 함수 연결
course_list = list(final_course_set)
print(f"\n총 {len(course_list)}개의 강의에 대해 강의평 크롤링을 진행합니다.")

# for name in course_list:
#     # 여기에 이전에 만드신 '응데분' 크롤링 함수명을 넣으세요.
#     # crawl_reviews(name) 
#     print(f"[{name}] 크롤링 대기 중...")

print("\n작업이 모두 종료되었습니다.")
```

```python
import pandas as pd
import time
import random
import re
import os
import urllib.parse
from bs4 import BeautifulSoup
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# [설정] 누적 저장할 파일명
final_output_file = 'khu_all_reviews_accumulated.csv'

# 강의 리스트 불러오기
df_names = pd.read_csv('khu_final_data_v1.csv', encoding='utf-8-sig')
course_list = df_names['강의명'].tolist()

for course_name in course_list:
    try:
        print(f"\n{'='*30}\n🔎 [{course_name}] 작업 시작...")
        
        # (1) URL 직접 검색 페이지 접속
        encoded_name = urllib.parse.quote(course_name)
        driver.get(f"https://everytime.kr/lecture/search?keyword={encoded_name}")
        
        # (2) 검색 결과가 나타날 때까지 대기 (최대 15초)
        wait = WebDriverWait(driver, 15)
        try:
            # 검색 결과 박스들이 나타날 때까지 대기
            wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "a.container, a.lecture")))
            results = driver.find_elements(By.CSS_SELECTOR, "a.container, a.lecture")
        except:
            print(f"❌ [{course_name}] 검색 결과가 화면에 뜨지 않았습니다.")
            continue

        # (3) 가장 알맞은 강의 매칭 및 클릭
        target_element = None
        for res in results:
            try:
                # h3 태그 내부의 강의명 텍스트 확인
                res_title = res.find_element(By.TAG_NAME, "h3").text.strip()
                if res_title == course_name:
                    target_element = res
                    break
            except: continue
        
        # 일치하는 게 없으면 첫 번째 결과 선택
        if not target_element and results:
            target_element = results[0]

        if target_element:
            # 일반 클릭이 안 될 경우를 대비해 자바스크립트로 강제 클릭
            driver.execute_script("arguments[0].click();", target_element)
            time.sleep(3)
        else:
            continue

        # (4) 강의평 탭(?tab=article) 이동
        if '?tab=article' not in driver.current_url:
            driver.get(driver.current_url + "?tab=article")
            time.sleep(3)

        # (5) 무한 스크롤 (리뷰 로딩)
        try:
            scroll_box = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, 'div.articles')))
            last_height = driver.execute_script("return arguments[0].scrollHeight", scroll_box)
            for _ in range(10): # 최대 10번 스크롤
                driver.execute_script("arguments[0].scrollTop = arguments[0].scrollHeight", scroll_box)
                time.sleep(1.5)
                new_height = driver.execute_script("return arguments[0].scrollHeight", scroll_box)
                if new_height == last_height: break
                last_height = new_height
        except: pass

        # (6) 데이터 추출
        soup = BeautifulSoup(driver.page_source, 'html.parser')
        
        # 올바른 강의명 및 전체 평점 추출
        # 상세 페이지 상단의 강의 제목(h2)을 가져옵니다.
        raw_title = soup.select_one('div.side h2').get_text(strip=True) if soup.select_one('div.side h2') else course_name
        real_name = re.sub(r'[\s\n\t\r]+', ' ', raw_title).strip()
        avg_score = soup.select_one('div.average').get_text(strip=True) if soup.select_one('div.average') else "0"

        reviews = soup.select('div.article')
        data_list = []
        for r in reviews:
            txt = r.select_one('div.text').get_text(strip=True)
            if txt:
                # 개별 평점(별점)
                star_score = 0
                star_on = r.select_one('span.on')
                if star_on and 'style' in star_on.attrs:
                    w = re.search(r'width:\s*(\d+)%', star_on['style'])
                    if w: star_score = int(w.group(1)) / 20
                
                sem = r.select_one('span.semester').get_text(strip=True) if r.select_one('span.semester') else "정보없음"
                
                data_list.append({
                    '강의명': real_name,
                    '전체평균평점': avg_score,
                    '개별평점': star_score,
                    '수강시기': sem,
                    '서술평': txt
                })

        # (7) 데이터 누적 저장
        if data_list:
            df_temp = pd.DataFrame(data_list)
            # 파일이 없으면 새로 생성, 있으면 아래에 추가
            if not os.path.exists(final_output_file):
                df_temp.to_csv(final_output_file, index=False, encoding='utf-8-sig')
            else:
                df_temp.to_csv(final_output_file, index=False, encoding='utf-8-sig', mode='a', header=False)
            print(f"✅ [{real_name}] {len(data_list)}개 누적 저장 완료!")
        
        time.sleep(random.uniform(4, 6))

    except Exception as e:
        print(f"❌ [{course_name}] 오류 발생: {e}")
        continue

print(f"\n🚀 작업 완료! 결과 파일: {final_output_file}")
```
